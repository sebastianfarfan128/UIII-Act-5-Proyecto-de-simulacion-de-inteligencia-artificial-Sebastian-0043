### Proyecto de simulacion de inteligencia artificial con if, else, elif y operadores

```python
print("--- Sistema Admin AI ---")
accion = input("¿Qué desea hacer? (estado / reporte / seguridad): ")

if accion == "estado":
    print("Estado del servidor: Operativo y estable.")
elif accion == "reporte":
    print("Generando reporte... No se detectaron anomalías hoy.")
elif accion == "seguridad":
    clave = input("Ingrese clave de administrador: ")
    if clave == "admin123" or clave == "root":
        print("Acceso concedido. Protocolos de seguridad activos.")
    else:
        print("Acceso denegado. Alerta enviada.")
else:
    print("Comando no reconocido. Intente de nuevo.")
print("Elaborado por Sebastian Farfan 0043 NL = 18")
```
